## Objectives

- Analyze distributions
- Find relationships between variables
- Identify important features

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\data\\StudentPerformanceFactors.csv")

df.head(10)
```

```python
print(df.shape)
display(df.describe().T)
display(df.head())
```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
#Correlation Heatmap
plt.figure(figsize = (12,8))

sns.heatmap(
    corr ,
    annot = True,
    cmap = "coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
#Distribution of exam scores
plt.figure(figsize = (8,5))

sns.histplot(df["Exam_Score"], bins= 20 ,kde=True ,color = "green")

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize = (8,5))

sns.scatterplot(
    data = df ,
    x = "Hours_Studied",
    y = "Exam_Score",
    color = "red"


)

plt.title("Study hours vs Exam Score")
plt.show()
```

```python
#Attendence vs Exam Score
plt.figure(figsize=(8,5))

sns.scatterplot(
    data = df,
    x = "Attendance",
    y = "Exam_Score",
    color = "orange"
)

plt.title("Attendence vs Exam Score")

plt.show()
```

```python
#Boxplot
plt.figure(figsize=(8,5))

sns.boxplot(y=df["Exam_Score"])

plt.title("Exam Score Boxplot")

plt.show()
```

```python
plt.figure(figsize = (8,5))

sns.countplot(
    data = df,
    x = "Gender",

    color ="skyblue",
    width = (0.2,0.2)
)

plt.title("Gender Distribution")
plt.show()
```

```python
#Average Score by Gender
average = df.groupby("Gender")["Exam_Score"].mean()
average
```

```python
#Visualization
average.plot(kind = "bar",color = ["pink","skyblue"])

plt.ylabel("Average Score")
plt.title("Average Score by Gender")
plt.show()
```

```python
#Average score by parental education
plt.figure(figsize = (10,8))

sns.barplot(
    data = df,
    x = "Parental_Education_Level",
    y = "Exam_Score",
    color = "orange"
    )

plt.xticks(rotation = 30)
plt.show()
```

```python
#pair plot
sns.pairplot(
    df[
        [
            "Hours_Studied",
            "Attendance",
            "Sleep_Hours",
            "Previous_Scores",
            "Exam_Score"
        ]
    ]
)

plt.show()
```
