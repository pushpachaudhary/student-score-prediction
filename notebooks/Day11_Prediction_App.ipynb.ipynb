# Day 11 – Building Student Score Prediction App

## Objectives

- Train the model
- Take user input
- Predict the exam score
- Display the prediction
- Save the trained model

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

import joblib
```

```python
df = pd.read_csv("C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\data\\student_performance_processed.csv")

df.head()
```

```python
target = "Exam_Score"

X = df.drop(columns=[target])

Y = df[target]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
model = LinearRegression()
model.fit(X_train,Y_train)
print("Model trained successfully!")
```

```python
joblib.dump(model,"C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\models\\student_score_model.pkl")

print("Model saved successfully!")
```

```python
loaded_model = joblib.load("C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\models\\student_score_model.pkl")
```

```python
sample_student = X_train.iloc[[0]]

sample_student
```

```python
prediction = loaded_model.predict(sample_student)
print("predicted exam score : ",round(prediction[0],2))
```

```python
actual_score = Y_test.iloc[0]

print("Actual Score : ", actual_score)
print("Predicted Score :", round(prediction[0],2))
```

```python
predictions = loaded_model.predict(X_test.head(5))

results = pd.DataFrame({
    "Actual Score": Y_test.head(5).values,
    "Predicted Score": np.round(predictions, 2)
})

results
```

```python
results.to_csv("C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\data\\final_predictions.csv", index=False)

print("Predictions saved successfully!")
```
