# Day 9 – Train-Test Split & Linear Regression

## Objectives

- Separate Features & Target
- Split the data into training and testing sets
- Train a Linear Regression model
- Make predictions
- Compare actual and predicted values
- Understand how the model learns

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\data\\StudentPerformanceFactors.csv")

df.head()
```

```python
print("Shape :", df.shape)

display(df.head())

df.info()
```

```python
#separating Features & target
target = "Exam_Score"

X = df.drop(columns =[target])

Y = df[target]
```

```python
print("Features :", X.shape)
print("Target :", Y.shape)
```

```python
#Train Test Split
X_train , X_test, Y_train ,Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
print("Training Data :",X_train.shape)
print("Testing Data :", X_test.shape)
```

```python
#creating model
model = LinearRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
#make predictions
Y_pred = model.predict(X_test)
```

```python
#compare predictions
comparison = pd.DataFrame({
    "Actual Score " : Y_test.values,
    "Predicted Score " : Y_pred
})

comparison.head(10)
```

```python
#Display model coefficient
coefficients = pd.DataFrame({
    "Features ": X.columns,
    "Coefficient": model.coef_
})

coefficients.sort_values(
    by="Coefficient",
    ascending=False
)
```

```python
#intercept
print("Intercept :",model.intercept_)
```

```python
#saving prediction results
comparison.to_csv(
    "C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\data\\StudentPerformanceFactors.csv",
    index=False
)

print("Prediction results saved")
```
