```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("C:\\Users\\pushp\\OneDrive\\Desktop\\AI &ML project\\student-score-prediction-1\\data\\StudentPerformanceFactors.csv")
```

```python
df.head()
```

```python
#Dataset overview
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

display(df.head())
```

```python
#Identify Target & Features

target = "Exam_Score"
X = df.drop(columns = [target])
Y = df[target]

print("Target Variable : ", target)
print("Number of Features : ", X.shape[1])
```

```python
#categorize Features

numerical_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(include = "object").columns.tolist()


print("Numerical Features :")
print(numerical_features)

print("\nCategorical_features : ")
print(categorical_features)
```

```python
#correlation with Target
correlation = df.corr(numeric_only = True)
correlation["Exam_Score"].sort_values(ascending=False)
```

```python
#Create New Features
#study efficiency
df["Study_Efficiency"] = (
    df["Hours_Studied"] * df["Attendance"]
) / 100
```

```python
#Academic consistency
df["Academic_Consistency"] = (
    df["Previous_Scores"] + df["Exam_Score"]
) /2
```

```python
#healthy lifestyle index
df["Healthy_Lifestyle_Index"] = (
    df["Sleep_Hours"] + df["Physical_Activity"]
)/2
```

```python
df.head()
```

```python
#Correlation After Feature Engineering
corr = df.corr(numeric_only= True)

corr["Exam_Score"].sort_values(ascending=False)
```

```python
#Check for Constant Columns

constant_columns = [
    col for col in df.columns
    if df[col].nunique() == 1
]

print("Constant Columns:")
print(constant_columns)
```

```python
#Check Unique Values
for col in categorical_features:
    print("=" * 50)
    print(f"{col}")
    print(df[col].value_counts())
```
